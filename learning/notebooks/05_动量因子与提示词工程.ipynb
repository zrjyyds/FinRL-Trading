# 05 动量因子与提示词工程

## 本课学习目标

- A. 量化金融主线：动量因子（Momentum Factor）
零基础解释：用过去一段时间涨跌幅衡量资产强弱。
- B. 大语言模型主线：提示词工程（Prompt Engineering）和温度（Temperature）
零基础解释：提示词越清晰，输出越容易被程序使用；Temperature 控制随机性。
- C. 两条线如何连接：把市场数据和新闻文本转成可检查的表格信号。
- D. 可运行实验：计算 20 日动量，比较模糊和严格提示词。
- E. 结果解释：观察表格、图表和结构化输出。
- F. 常见错误：把回测收益当成未来收益、把 Mock 当成真实模型。
- G. 课后练习：修改一个参数并重新运行。
- H. 本课术语表：见本课各小节。
