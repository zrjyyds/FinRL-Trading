## 结尾总结

你现在应该理解：动量因子是量化的基础工具，提示词工程决定了模型输出的可用性。

本课核心收获：
- 动量因子用过去一段时间的涨跌幅衡量资产强弱；
- 排名加权将动量值转成统一的选股分数；
- 提示词不是简单的问题——它包含任务、格式、约束和示例；
- 模糊提示词让模型自由发挥，明确提示词让输出可被程序解析；
- Temperature 控制输出随机性：0 表示确定性最强，越高越有"创意"。

哪些结果不能解释为策略一定赚钱：任何图表和收益数字都只是合成数据上的教学结果。

本课使用了哪些英文专业词：
- Momentum Factor（动量因子）
- Prompt Engineering（提示词工程）
- Temperature（温度）
- Look-back Period（回看周期）

### 常见错误

1. **回看周期选择随意**：20 日动量捕捉的是短期趋势，改用 5 日或 60 日会得到完全不同的排名。
2. **把 Prompt 当成代码**：提示词约束模型但不强制——模型仍可能输出不符合格式的回答。
3. **Temperature=0 不等于"完美答案"**：Temperature 只控制随机性，不保证正确性。
4. **忘记标准化动量**：不同股票的动量绝对值不可直接比较，必须排名或标准化。

### 课后练习

1. **改变回看周期**：将 `compute_momentum(prices, 20)` 改为 10 日和 60 日，分别运行，观察排名前三的股票是否不同。
2. **分析排名变化**：短周期（10日）和长周期（60日）的动量排名有什么差异？哪种更适合捕捉趋势？
3. **改进提示词**：在 strict_prompt 中增加一个 "示例" 字段，写一个完整的 SentimentResult JSON 作为 few-shot 示例。
